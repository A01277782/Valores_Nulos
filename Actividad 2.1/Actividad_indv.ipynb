```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Citas_Digital_Filtrado.csv')
df.head(5)
```

```python
#Se eliminaron columnas innecesarias (Unnamed:15)
df1= df.drop(['Unnamed: 15'], axis=1)
df1
```

```python
valores_nulos = df.isnull().sum()
valores_nulos
```

```python
df2 = df1.copy()
```

```python
#Se sustituyeron los valores nulos por valores no nulos hacia adelante 
df2["BDC "] = df2["BDC "].ffill()
```

```python
valores_nulos = df2.isnull().sum()
valores_nulos
```

```python
#Se sustiuyo valores nulos por una fecha ficticia
df2["Fecha"] =df2["Fecha"].fillna("2030-06-19 00:00:00") 
```

```python
valores_nulos = df2.isnull().sum()
valores_nulos
```

```python
#Se sustituyeron los valores nulos por valores no nulos hacia adelante 
df2["Nombre Cliente "] = df2["Nombre Cliente "].ffill()
```

```python
valores_nulos = df2.isnull().sum()
valores_nulos
```

```python
#se sustituyeron los valores nulos de telefono por texto de "no identificado"
df2["Telefono "] =df2["Telefono "].fillna("No identificado") 
```

```python
valores_nulos = df2.isnull().sum()
valores_nulos
```

```python
#Se sustituyeron los valores nulos por valores no nulos hacia adelante 
df2["Estatus de Lead"] = df2["Estatus de Lead"].ffill()
```

```python
valores_nulos = df2.isnull().sum()
valores_nulos
```

```python
#se cambiaron los valores a float
df2['Potencial de compra'].unique

df2['Potencial de compra']=df2['Potencial de compra'].astype("string")
df2['Potencial de compra']=df2['Potencial de compra'].str.replace("Ventas ","0", regex=False)
df2['Potencial de compra']=df2['Potencial de compra'].str.replace("%","", regex=False)
df2['Potencial de compra']=df2['Potencial de compra'].str.replace("'","", regex=False)
df2['Potencial de compra']=df2['Potencial de compra'].str.replace("&","", regex=False)

df2['Potencial de compra']=df2['Potencial de compra'].astype(float)
```

```python
#Primer método de sustitución de valores nulos
#Sustituir valores nulos con promedio o media
df2["Potencial de compra"]=df2["Potencial de compra"].fillna(round(df2["Potencial de compra"].mean(),1))
```

```python
valores_nulos = df2.isnull().sum()
valores_nulos
```
